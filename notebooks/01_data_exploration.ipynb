Contract แบบไหน Churn เยอะที่สุด?

```python
contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn Label"],
    normalize="index"
) * 100

contract_churn
```

```python
contract_churn.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Insight: Contract Type

Customers with month-to-month contracts have a substantially higher
churn rate (42.7%) compared with customers on one-year (11.3%)
and two-year contracts (2.8%).

This suggests that customers with shorter-term contracts may have
higher churn risk and could be an important segment for retention
strategies.

Monthly Charges สูงขึ้น ลูกค้า Churn มากขึ้นไหม?

```python
sns.boxplot(
    data=df,
    x="Churn Label",
    y="Monthly Charges"
)

plt.title("Monthly Charges by Churn Status")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")
plt.show()
```

### Insight: Monthly Charges

Customers who churn tend to have higher monthly charges than
customers who remain with the company.

The median monthly charge for churned customers is approximately
$80, compared with approximately $65 for non-churned customers.

This suggests that monthly charges may be associated with customer churn.

ลูกค้าที่อยู่กับบริษัทนานกว่า Churn น้อยลงไหม?

```python
sns.boxplot(
    data=df,
    x="Churn Label",
    y="Tenure Months"
)

plt.title("Tenure by Churn Status")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")
plt.show()
```

### Insight: Customer Tenure

Customers who churn tend to have much shorter tenure than customers
who remain with the company.

The median tenure is approximately 10 months for churned customers,
compared with approximately 38 months for non-churned customers.

This indicates that early-stage customers may represent a higher-risk
segment for churn.

Internet Service แบบไหนมี Churn สูง?

```python
internet_churn = pd.crosstab(
    df["Internet Service"],
    df["Churn Label"],
    normalize="index"
) * 100

internet_churn
```

```python
internet_churn.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Churn Rate by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Insight: Internet Service

Fiber optic customers have the highest churn rate at approximately
41.9%, compared with 19.0% for DSL customers and 7.4% for customers
without internet service.

This makes internet service type a potentially important feature
for churn prediction.
